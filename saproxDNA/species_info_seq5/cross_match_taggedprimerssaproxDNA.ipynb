```python
mft_primers_path = '../../data/saproxdna_primertags_seq5_2026_1.fasta'

header = []
seq = []
with open(mft_primers_path, 'r', newline='', encoding='utf-8') as file:
    for line in file:
        if line.startswith('>'):
            header.append(line.strip().removeprefix('>'))
        else:
            seq.append(line.strip())

mft_primers = dict(zip(header,seq))
mft_primers
```

```python
pcr_primers_path = '../../data/PCRsaproxdna_primertags_seq5_2026_1.fasta'

header = []
seq = []
with open(pcr_primers_path, 'r', newline='', encoding='utf-8') as file:
    for line in file:
        if line.startswith('>'):
            header.append(line.strip().removeprefix('>'))
        else:
            seq.append(line.strip())

pcr_primers = dict(zip(header,seq))
pcr_primers
```

```python
# for key,value in mft_primers.items():
#     if value in pcr_primers.values():
#         print(pcr_primers)


pcr_by_value = {value: key for key, value in pcr_primers.items()}
mft_to_pcr_names = {}
for key, value in mft_primers.items():
    if value in pcr_by_value.keys():
        print('PCR name: ', pcr_by_value[value], 'MFT name: ', key)
        mft_to_pcr_names[key] = pcr_by_value[value]
    else:
        print('Not in PCR primers: ',key, value)
mft_to_pcr_names
```
